### C1. Load records safely from a file

Write `load_records(path)` with a one-line docstring. It reads `label,value,limit` lines from a file
and returns a **list of dicts**. If the file does not exist, it prints a clear message and returns an
empty list instead of crashing.

Then load `records_sample.txt` and print each record's label and status, as in D16.

**Test it:**

| Call | Expected output |
|---|---|
| `load_records("records_sample.txt")` | the same three lines as D16 |
| `load_records("nope.txt")` | a clear message, then `[]` |

````python
#Define a function which is called load_records
def load_records(path):
    """Read records and displays information in terms of data that is split in value and limit."""
    #Check if condition is met
    try:
        #Create a variable which is called new_records
        new_records = []
        #Open a file 
        #f is assigned the variable for the opening of the file
        with open (path) as f:
            #loop to seperate line
            for line in f:
                #Split values on different label types using , as reference point
                current_label, current_value, current_limit = line.strip().split(",")
                #Convert current_value to float
                current_value = float(current_value)
                #Convert current_limit to float
                current_limit = float(current_limit)
                #Add label, value and limit to new_records
                new_records.append({"label": current_label, "value": current_value, "limit": current_limit})
            #loop to seperate individual_record from new_records
            for individual_record in new_records:
                #Create a variable which is called Status and check if the one of two conditions is met
                Status = "OVERLIMIT" if individual_record["value"] > individual_record["limit"] else "OK"
                #Display label and status from indiviudal_record
                print(f"{individual_record["label"]}: {Status}")
    #handling of exception
    except FileNotFoundError:
        #Display a clear message
        print("The file does not exist.")
    #Return neew_records to a user
    return new_records
#Call a function which is called load_records with contains a parameter named records_sample.txt for the first time
load_records("records_sample.txt")
#Call a function which is called load_records with contains a parameter named nope.txt for the second time
load_records("nope.txt")
````

### C2. Write the results to a file

Using your records from C1, write one line per record to a new file called `results.txt`,
in the form `label,status`. Then read `results.txt` back and print it.

**Expected output:**
```
R-001,OK
R-002,OVER LIMIT
R-003,OK
```

````python
#Run C1 in order to recognise a function load_records()
#Create a record and assign a function to it
records = load_records("records_sample.txt")
"""For each record, opening a file and assigning status."""
#Opening file in write mode
#f is assigned the variable for the opening of the file
with open("results.txt", "w") as f:
    #loop to seperate a record from records
    for record in records:
        #Create a variable which is called Status and check if the one of two conditions is met
        Status = "OVERLIMIT" if record["value"] > record["limit"] else "OK"
         #Create a variable which is called status and check if the one of two conditions is met
        record["Status"] = Status
        #Write a label and status from a record
        f.write(f"{record["label"]},{record["Status"]}\n")
#Opening file in write mode
#f is assigned the variable for the opening of the file
with open("results.txt") as f:
    #Display everything that is contained from results.txt
    print(f.read())
````
