### D6. `self` is whichever object made the call

1. Copy your class from D5 and add a method `status(self)` that returns `"OVER LIMIT"` (100% or more),
   `"WARNING"` (90% or more) or `"OK"`, worked out from `self.value` and `self.limit`.
2. Create three records and print each one's `.status()`:

| Record | Expected output |
|---|---|
| `Record("R-001", 87.0, 100.0)` | `OK` |
| `Record("R-002", 120.0, 100.0)` | `OVER LIMIT` |
| `Record("R-003", 95.0, 100.0)` | `WARNING` |

````python
"""Create a record class."""
#Define a class named record
class Record:
    #Define constructor
    def __init__(self, label, value, limit):
        self.label = label
        self.value = value
        self.limit = limit
    #Define a method which is called status(self)
    def status(self):
        #Calculate the percent of value over limit times 100
        percent = (self.value / self.limit) * 100
        #Return appropriate message based on percent
        return "OVER LIMIT" if percent >= 100 else "WARNING" if percent >= 90 else "OK"
#Create three instances of record
r = Record("R-001", 87.0, 100.0)
r1 = Record("R-002", 120.0, 100.0)
r2 = Record("R-002", 95.0, 100.0)
#Display message by calling status method
print(r.status())
print(r1.status())
print(r2.status())
````

### D7. One method calls another

1. Copy your class from D6 and add a method `percent(self)` that returns `(self.value / self.limit) * 100`.
2. Change `status(self)` so it uses `self.percent()` instead of working out the percentage itself.
3. Create `Record("R-001", 87.0, 100.0)` and print its `.percent()` and `.status()`.

**Expected output:** `87.0`, then `OK`

````python
"""Create a record class."""
#Define a class named record
class Record:
    #Define constructor
    def __init__(self, label, value, limit):
        self.label = label
        self.value = value
        self.limit = limit
    #Define a method which is called percent(self)
    def percent(self):
        #Return percentage of value divide by limit times 100
        return (self.value/ self.limit) * 100
    #Define a method which is called status(self)
    def status(self):
        #Return appropriate message based on percent
        return "OVER LIMIT" if self.percent() >= 100 else "WARNING" if self.percent() >= 90 else "OK"
#Create an instance of record
r = Record("R-001", 87.0, 100.0)
#Display percentage by calling percent method
print(r.percent())
#Display status by calling status method
print(r.status())
````

### D8. A method can change its own object

1. Define a class `Checker`. In `__init__(self)`, set `self.over_limit_count = 0`.
2. Add a method `check(self, value, limit)` that adds 1 to `self.over_limit_count` when `value` is
   greater than `limit`.
3. Create one `Checker`, call `.check()` three times with `(87, 100)`, `(120, 100)` and `(130, 100)`,
   then print `over_limit_count`.

**Expected output:** `2`

````python
"""Create a new class of checker."""
#Define a class named Checker
class Checker:
    #Define constructor
    def __init__(self):
        #Initialise over_limit_count
        self.over_limit_count = 0
    #Define a method which is called check(self, value, limit)
    def check(self, value, limit):
        #Check if value is greater than limit
        if value > limit:
            #over_limit_count is incremented by 1
            self.over_limit_count += 1
#Creates an object for checker
c = Checker()
#Call check method
c.check(87, 100)
c.check(120, 100)
c.check(130, 100)
#Display over_limit_count
print(c.over_limit_count)
    
````

### D11. Load objects from a file

`records_sample.txt` (in this `LAB` folder) has one record per line, like `R-001,87,100`.

1. Write `load_records(path)` that reads the file and returns a list of `Record` objects
   (convert `value` and `limit` with `float()`).
2. Loop over the list and print each record's label and status.

**Expected output:** the same three lines as D9.

````python
#Define a function which is called load_records(path)
def load_records(path):
    """Turn file into a list data of record objects."""
    #Define a list of record
    records = []
    #Open the file
    #f is assigned the variable for the opening of the file
    with open(path) as f:
       #loop to read each line and add it to the list of records
       for line in f:
            label, value, limit = line.strip().split(",")
            records.append(Record(label, float(value), float(limit)))
    #loop to seperate record from records
    for record in records:
        #Display label and status
        print(record.label, record.status())
    #Return the list of records
    return records
#Call load_records using file
localrecord = load_records("records_sample.txt")
````

### C2. An account that locks

Write a class `Account` with a `username` and a count of failed logins that starts at 0.
It has two methods:

- `fail(self)`: adds 1 to the failed logins
- `locked(self)`: returns `True` if there have been 3 or more failed logins

**Test it:**

| Code | Expected output |
|---|---|
| an account for `"anna"` with `.fail()` called 3 times, then `print(anna.locked())` | `True` |
| an account for `"ben"` with `.fail()` called once, then `print(ben.locked())` | `False` |

````python
# Create class account
class Account:
   # Define constructor
    def __init__(self, username):
        self.username = username
        self.failed_logins = 0
   
   # Define fail method 
    def fail(self):
        """Add 1 to the number of failed logins."""
        self.failed_logins += 1
   # Define locked method
    def locked(self):
        """Return True if there have been 3 or more failed logins."""
        return self.failed_logins >= 3

#Create an instance for account "anna"
anna = Account("anna")


# for loop 3 times
for a in range(3):
   # Call fail method for failed logging in has failed
    anna.fail()
#lock anna since she has failed to log in 3 times
print(anna.locked())    # True

#Create an instance for account "Ben"
ben = Account("ben")
# Call fail to log in once
ben.fail()
#do not lock ben since he has only failed once
print(ben.locked())     # False
````
